## IRIS Dataset을 통한 Randomforest pipeline 구축

```python
from sklearn import datasets
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

x, y = datasets.load_iris(return_X_y=True)
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=0)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

pipe = make_pipeline(
    StandardScaler(),
    RandomForestClassifier()
)
```

```python
pipe.fit(x_train, y_train)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, pipe.predict(x_test))
```

```python
pipe2 = make_pipeline(
    StandardScaler(),
    linear_model.LogisticRegression()
)
pipe2.fit(x_train, y_train)
```

```python
from sklearn.metrics import top_k_accuracy_score
print(accuracy_score(y_test, pipe2.predict(x_test)))
print(top_k_accuracy_score(y_test, pipe2.predict_proba(x_test), k=2))
```

## Isomap

```python
from sklearn.manifold import Isomap
isomap = Isomap(n_neighbors=20)
embedding = isomap.fit_transform(X=x_train, y=y_train)
```

```python
embedding.shape
```

```python
plt.figure(figsize=(5,5))
plt.scatter(x=embedding[:, 0], y=embedding[:, 1], c=y_train)
plt.show()
```

```python
pipe3 = make_pipeline(
    linear_model.LogisticRegression()
)
pipe3.fit(X=embedding, y=y_train)
test_embedding = isomap.transform(X=x_test)
accuracy_score(y_test, pipe3.predict(test_embedding))
```

```python
pipe4 = make_pipeline(
    Isomap(n_neighbors=20, n_components=2),
    linear_model.LogisticRegression()
)
pipe4.fit(X=x_train, y=y_train)
print(accuracy_score(y_test, pipe4.predict(x_test)))
print(top_k_accuracy_score(y_test, pipe2.predict_proba(x_test), k=2))
```
